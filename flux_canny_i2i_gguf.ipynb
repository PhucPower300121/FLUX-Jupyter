```python
#@title Load model to RAM (GGUF, mmap -- low RAM)
%cd /content/ComfyUI

import random, torch, numpy as np
from PIL import Image
from nodes import NODE_CLASS_MAPPINGS
import nodes

# Nạp toàn bộ node built-in (comfy_extras, gồm cả node Canny) + custom node (GGUF...) qua cơ chế chính thức của ComfyUI
from nodes import init_extra_nodes
await init_extra_nodes()

UnetLoaderGGUF = NODE_CLASS_MAPPINGS["UnetLoaderGGUF"]()
DualCLIPLoader = NODE_CLASS_MAPPINGS["DualCLIPLoader"]()
VAELoader = NODE_CLASS_MAPPINGS["VAELoader"]()
CLIPTextEncode = NODE_CLASS_MAPPINGS["CLIPTextEncode"]()
KSampler = NODE_CLASS_MAPPINGS["KSampler"]()
VAEDecode = NODE_CLASS_MAPPINGS["VAEDecode"]()
LoadImage = nodes.LoadImage()
InstructPixToPixConditioning = NODE_CLASS_MAPPINGS["InstructPixToPixConditioning"]()
Canny = NODE_CLASS_MAPPINGS["Canny"]()

with torch.inference_mode():
    clip = DualCLIPLoader.load_clip("t5xxl_fp8_e4m3fn.safetensors", "clip_l.safetensors", "flux")[0]
    unet = UnetLoaderGGUF.load_unet("flux1-canny-dev-Q4_0.gguf")[0]
    vae = VAELoader.load_vae("ae.sft")[0]

print("Model load complete.")
```

```python
#@title Upload photos
from google.colab import files
uploaded = files.upload()
import shutil, os
init_image_name = list(uploaded.keys())[0]
dst = os.path.join("/content/ComfyUI/input", init_image_name)
shutil.copy(init_image_name, dst)
print("Uploaded:", init_image_name)
```

```python
#@title Run Canny-guided Generation
positive_prompt = "a beautiful landscape, high detail, cinematic lighting"  #@param {type:"string"}
negative_prompt = ""  #@param {type:"string"}
low_threshold = 0.4  #@param {type:"slider", min:0.0, max:1.0, step:0.01}
high_threshold = 0.8  #@param {type:"slider", min:0.0, max:1.0, step:0.01}
steps = 20  #@param {type:"slider", min:10, max:50, step:1}
cfg = 10  #@param {type:"number"}
sampler_name = "euler"  #@param ["euler", "euler_ancestral", "dpmpp_2m", "dpmpp_2m_sde"]
scheduler = "simple"  #@param ["simple", "normal", "karras"]
seed = 0  #@param {type:"integer"}

# Canny-dev là model input-channel riêng (không phải ControlNet apply thường):
# edge map được ghép trực tiếp vào conditioning qua InstructPixToPixConditioning.
# cfg khuyến nghị cao hơn t2i thường (~10) để bám sát edge map.

with torch.inference_mode():
    if seed == 0:
        seed = random.randint(0, 18446744073709551615)
    print("Seed:", seed)

    # Tạo canny edge map từ ảnh gốc bằng node Canny built-in
    ref_image = LoadImage.load_image(init_image_name)[0]
    canny_image = Canny.detect_edge(ref_image, low_threshold, high_threshold)[0]
    # Canny trả ảnh 1 kênh -> nhân bản thành 3 kênh để encode qua VAE như ảnh RGB
    canny_image = canny_image.repeat(1, 1, 1, 3)

    positive = CLIPTextEncode.encode(clip, positive_prompt)[0]
    negative = CLIPTextEncode.encode(clip, negative_prompt)[0]

    # Ghép edge map vào conditioning + tạo latent theo kích thước edge map
    positive, negative, latent_image = InstructPixToPixConditioning.encode(
        positive, negative, vae, canny_image
    )

    samples = KSampler.sample(
        unet, seed, steps, cfg, sampler_name, scheduler,
        positive, negative, latent_image, denoise=1.0
    )[0]

    decoded = VAEDecode.decode(vae, samples)[0].detach()
    out_img = Image.fromarray(np.array(decoded * 255, dtype=np.uint8)[0])
    out_img.save("/content/output.png")

out_img
```
